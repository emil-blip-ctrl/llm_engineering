```python
import sys
from pathlib import Path

repo_root = Path(r"C:/Users/ME36352/MyFolder/AI/Repos/AIRepo")
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from aceRag.retrieval_call import retrieval_call

```

```python
chunks = retrieval_call(
    "How to tune integration server? How to lower its starting time and debugger attaching time?"
)

output_path = repo_root / "codeGeneratorV3" / "retrieved_chunks.txt"
blocks = []
for index, chunk in enumerate(chunks, start=1):
    blocks.append(
        f"{'=' * 80}\n"
        f"Chunk {index} of {len(chunks)}\n"
        f"{'=' * 80}\n"
        f"{chunk.strip()}\n"
    )
output_path.write_text("\n".join(blocks), encoding="utf-8")
print(f"Saved {len(chunks)} chunks to {output_path}")
```
