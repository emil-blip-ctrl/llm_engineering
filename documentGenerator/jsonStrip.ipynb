```python
import json

# Read raw JSON file
with open("input.json", "r", encoding="utf-8") as f:
    data = json.load(f)

# Example:
# {
#   "content": "## Summary\\n\\nSome text..."
# }

# Extract markdown string
markdown_text = data["document"]

# Save as real markdown file
with open("output.md", "w", encoding="utf-8") as f:
    f.write(markdown_text)

print("Markdown extracted successfully.")
```
