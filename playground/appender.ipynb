```python
# Concatenate all files from a given directory into concatinatedContent.txt in the CWD

from pathlib import Path
import shutil

# Set the directory you want to read files from.
# Use "." for the current directory, or change to a specific path like "/path/to/dir"
directory = "C:/Users/ME36352/MyFolder/REPOS/ACE/finance-accounting-integration-technical-invoices-financial-documents-v2/apps/finance-accounting-integration-technical-invoices-financial-documents-v2/SAP"

source_dir = Path(directory).expanduser().resolve()
output_path = Path.cwd() / "xsdSchema.txt"

if not source_dir.is_dir():
    raise NotADirectoryError(f"Not a directory: {source_dir}")

# Collect and sort files by name (case-insensitive)
files = sorted((p for p in source_dir.iterdir() if p.is_file()), key=lambda p: p.name.lower())

written = 0
output_path.parent.mkdir(parents=True, exist_ok=True)

with output_path.open("w", encoding="utf-8", newline="\n") as out_f:
    for fp in files:
        # Skip the output file itself if it's in the same directory
        try:
            if fp.resolve() == output_path.resolve():
                continue
        except Exception:
            pass

        out_f.write(f"====={fp.name}=====\n\n")
        try:
            with fp.open("r", encoding="utf-8", errors="replace") as in_f:
                shutil.copyfileobj(in_f, out_f)
        except Exception as e:
            out_f.write(f"[Could not read file: {e}]\n")
        out_f.write("\n\n")
        written += 1

print(f"Wrote {written} file(s) into {output_path}")
```
