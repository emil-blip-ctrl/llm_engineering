```python
import requests
from bs4 import BeautifulSoup


def get_webpage_content(url: str, timeout: int = 30) -> str:
    """
    Fetch a webpage and return its cleaned text content.

    Args:
        url: The webpage URL.
        timeout: Request timeout in seconds.

    Returns:
        Cleaned text content from the page.

    Raises:
        requests.RequestException: If the request fails.
    """

    headers = {
        "User-Agent": (
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 (KHTML, like Gecko) "
            "Chrome/125.0 Safari/537.36"
        )
    }

    response = requests.get(url, headers=headers, timeout=timeout)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    # Remove unwanted elements
    for tag in soup([
        "script",
        "style",
        "noscript",
        "header",
        "footer",
        "nav",
        "aside"
    ]):
        tag.decompose()

    text = soup.get_text(separator="\n")

    # Clean whitespace
    lines = [line.strip() for line in text.splitlines()]
    lines = [line for line in lines if line]

    return "\n".join(lines)
```

```python
import asyncio
from playwright.async_api import async_playwright


async def get_page_content(url: str) -> str:
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()

        await page.goto(url, wait_until="networkidle", timeout=60000)
        await page.wait_for_timeout(3000)

        content = await page.inner_text("body")

        await browser.close()
        return content
```

```python
content = get_webpage_content("https://aiforgood.itu.int/summit26/programme/?")
print(content)
```

```python
user_prompt = """" I am giving you a web page content  """ + content + """ which contains 
different events/sessions.  By the content you need to figure out the events yourself. 
I want youy to list all and only the events accissible by "discovery" pass 
in the format date:time - title - type. In 24H time format. You have the date example
"Tuesday, 7 July" and below it all the events of this date, next you have for the next
day and the events for this day below it. So follow the events and list them in the format
 date:time - title - type. In 24H time format.
You have the content and you need to make the list.
Don't ask me questions back. This is API call. Not a conversation. You have the content, 
do the list! I want it as list, not JSON, not CSV or anything. Just a list. """
```

```python
def select_relevant_links():
    response = openai.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": ""},
            {"role": "user", "content": user_prompt}
        ]
    )
    result = response.choices[0].message.content
    return result
    
```

```python
print(select_relevant_links())
```
