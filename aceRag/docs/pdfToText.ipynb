```python
from typing import Any
import pymupdf
from pathlib import Path

# PDF file in the current working directory
pdf_path = Path("C:/Users/ME36352/Downloads/ace12-user_guide.pdf")

# Output text file in the same directory
txt_path = pdf_path.with_suffix(".txt")

if not pdf_path.exists():
    raise FileNotFoundError(f"PDF not found: {pdf_path.resolve()}")

with pymupdf.open(pdf_path) as pdf:
    with txt_path.open("w", encoding="utf-8") as txt_file:
        for page_number, page in enumerate(pdf, start=1):
            text = page.get_text("text")
            txt_file.write(text)

    print(f"PDF pages processed: {len(pdf)}")

print(f"Text file saved to: {txt_path.resolve()}")
```
